```python
import numpy as np
```

```python
a=10
b=20
numbers = [10,30,50,60,80]
```

```python
#Arithmetic Calculations
print("Addition of a & b :", a+b)
print("Subtraction of a & b :", a-b)
print("Multiplication of a & b :", a*b)
print("Division of a & b :", a/b)
print("Modulus of a & b :", a%b)
```

```python
#Statistical Calculations
print("Mean of numbers:", np.mean(numbers))
print("Median of numbers:", np.median(numbers))
print("Standard Deviation of numbers:", np.std(numbers))
```
