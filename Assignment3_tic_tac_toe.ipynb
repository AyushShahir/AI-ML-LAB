```python
board =[" " for _ in range (9)]

def print_board():
  print()
  print(board[0], "|", board[1], "|", board[2])
  print("-----------------")
  print(board[3], "|", board[4], "|", board[5])
  print("-----------------")
  print(board[6], "|", board[7], "|", board[8])
  print()

print_board()


```

```python
def check_winner(symbol):
  winning_position=[
      [0,1,2],[3,4,5],[6,7,8],
      [0,3,6],[1,4,7],[2,5,8],
      [0,4,8],[2,4,6]
  ]

  for position in winning_position:
    if board[position[0]]==symbol and board[position[1]]==symbol and board[position[2]]==symbol:
      return True
  return False
```

```python
#ai decision making process

def ai_turn():
  #check for win
  for i in range(9):
    if board[i]==" ":
      board[i]="O"
      if check_winner("O"):
        return
      board[i]=" "
  #check for block
  for i in range(9):
    if board[i]==" ":
      board[i]="X"
      if check_winner("X"):
        board[i]="O"
        return
      board[i]=" "
  #check for play
  for i in range(9):
    if board[i]== " ":
      board[i]="O"
      return
  #check for middle
  if board[4]==" ":
    board[4]="O"
    return
  #check for corners
  corners=[0,2,6,8]
  for i in corners:
    if board[i]==" ":
      board[i]="O"
      return
```

```python
#The game starts
print("X your symbol")
print("O AI's symbol")
print("Let's start tic-tac-toe")
print("You start!")
print_board()

for turn in range(9):

  if turn %2==0:

    while True:
       move=int(input("Enter your move (1-9): "))-1
       if move >= 0 and move < 9 and board[move]==" ":
         board[move]="X"
         break
       print("Invalid move! Try again.")
    print_board()

    if check_winner("X"):
      print("You win!")
      break

    else:
      ai_turn()
      print_board()
      if check_winner("O"):
          print("AI Wins!")
          break
else:
    print("It's a Draw!")
```
